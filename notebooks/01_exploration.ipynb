Imports

```python
import warnings
import pandas as pd
from statsbombpy import sb

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
```

Available competitions

```python
competitions = sb.competitions()
competitions[["competition_id", "season_id", "competition_name", "season_name"]]
```

Competitions matches

```python
wc = competitions[competitions["competition_name"] == "FIFA World Cup"]
print(wc[["competition_id", "season_id", "season_name"]])

matches = sb.matches(competition_id=43, season_id=106)
print(f"Nombre de matchs : {len(matches)}")
matches[["match_id", "home_team", "away_team", "home_score", "away_score"]].head()
```

Match events and shots

```python
match_id = matches["match_id"].iloc[0]
events = sb.events(match_id=match_id)
print(f"{len(events)} événements dans ce match")

shots = events[events["type"] == "Shot"]
print(f"{len(shots)} tirs")

cols = ["player", "team", "location", "shot_body_part", "shot_type",
        "play_pattern", "under_pressure", "shot_outcome", "shot_statsbomb_xg"]
shots[cols]
```
